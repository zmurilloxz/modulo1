```python
# Bibliotecas usadas na análise
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
# Estilo visual padrão dos gráficos
sns.set_theme(style='whitegrid')
```

```python
# Paleta de cores usada em todos os gráficos
COR_PRINCIPAL = '#2E86AB'
COR_DESTAQUE = '#E63946'
paleta = ['#2E86AB', '#A23B72', '#F18F01', '#C73E1D', '#3B1F2B', '#6C757D']

sns.set_palette(paleta)
```

```python
# Lê o arquivo, remove a coluna de índice antigo e mostra as primeiras linhas
df = pd.read_csv('ecommerce_preparados.csv')
df = df.drop(columns=['Unnamed: 0'])
print('Linhas e colunas:', df.shape)
df.head()
```

```python
# Tipos das colunas e quantidade de valores preenchidos
df.info()
```

Histograma e Densidade — Distribuição dos preços e das notas

```python
# Histograma dos preços e densidade das notas, lado a lado
fig, eixos = plt.subplots(1, 2, figsize=(14, 5))

eixos[0].hist(df['Preço'].dropna(), bins=40, color=COR_PRINCIPAL, edgecolor='white')
eixos[0].set_title('Distribuição dos Preços')
eixos[0].set_xlabel('Preço (R$)')
eixos[0].set_ylabel('Quantidade de produtos')

sns.kdeplot(df['Nota'].dropna(), fill=True, color=COR_PRINCIPAL, ax=eixos[1])
eixos[1].set_title('Densidade das Notas')
eixos[1].set_xlabel('Nota (1 a 5)')
eixos[1].set_ylabel('Densidade')

plt.tight_layout()
plt.show()
```

O que observar: a maior parte dos produtos fica entre R$ 50 e R$ 140, e poucos passam de R$ 300.

O que observar: as notas se concentram entre 4 e 5. Notas baixas são raras.

Dispersão - Relação entre Preço e Nota

```python
# Dispersão: cada ponto é um produto (preço x nota)
plt.figure(figsize=(10,6))

plt.scatter(df['Preço'], df['Nota'], alpha=0.4, color=COR_PRINCIPAL)

plt.title('Relação entre Preço e Notas dos Produtos')
plt.xlabel('Preço (R$)')
plt.ylabel('Nota (1 a 5)')
plt.show()
```

O que observar: a nuvem não tem direção. Preço caro não garante nota melhor.

Mapa de calor - Correlação entre as Variáveis Numéricas

```python
# Mapa de calor da correlação entre as variáveis numéricas
colunas = ['Nota', 'N_Avaliações', 'Desconto', 'Preço', 'Qtd_Vendidos_Cod', 'Marca_Freq']

correlacao = df[colunas].corr()

plt.figure(figsize=(9, 7))

escala = sns.blend_palette([COR_DESTAQUE, 'white', COR_PRINCIPAL], as_cmap=True)

sns.heatmap(correlacao, annot=True, fmt='.2f', cmap=escala, vmin=-1, vmax=1)

plt.title('Correlação entre as Variáveis Numéricas')
plt.xlabel('Variáveis comparadas')
plt.ylabel('Variáveis de referência')
plt.show()
```

O que observar: N_Avaliações x Qtd_Vendidos_Cod = 0,90, a relação mais forte da tabela. Produtos que vendem mais acumulam mais avaliações.

Barras - Top 10 Marcas com Mais Produtos

```python
# Barras: as 10 marcas com mais produtos (sem as marcas "não definido")
top_marcas = df[df['Marca'] != 'não definido']['Marca'].value_counts().head(10)

plt.figure(figsize=(10,6))

plt.barh(top_marcas.index[::-1], top_marcas.values[::-1], color=COR_PRINCIPAL)

plt.title('Top 10 Marcas com Mais Produtos Anunciados')
plt.xlabel('Quantidade de Produtos')
plt.ylabel('Marca')
plt.show()
```

O que observar: Lupo e Stillger lideram com folga.

Pizza - Distribuição dos Produtos por Gênero

```python
# Pizza: os 6 gêneros principais e o restante agrupado em "Outros"
generos = df['Gênero'].value_counts()

principais = generos.head(6).copy()
principais['Outros'] = generos.iloc[6:].sum()

pct = principais / principais.sum() * 100
rotulos = [f'{nome} ({p:.1f}%)' for nome, p in zip(principais.index, pct)]

plt.figure(figsize=(8, 8))

plt.pie(principais, labels=rotulos, startangle=90,
        colors=paleta + ['#ADB5BD'], wedgeprops={'edgecolor': 'white'})

plt.title('Distribuição dos Produtos por Gênero')
plt.show()
```

O que observar: Feminino 34%, Masculino 26%, Bebês 15%.

Regressão - Relação entre Avaliações e Vendas

```python
# Regressão: relação entre número de avaliações e quantidade vendida
plt.figure(figsize=(10,6))

sns.regplot(x='N_Avaliações', y='Qtd_Vendidos_Cod', data=df, scatter_kws={'alpha': 0.4, 'color': COR_PRINCIPAL}, line_kws={'color': COR_DESTAQUE})

plt.title('Relação entre Número de Avaliações e Quantidade Vendida')
plt.xlabel('Número de Avaliações')
plt.ylabel('Quantidade Vendida')
plt.show()
```

O que observar: a linha sobe bem, o que confirma o 0,90 visto no mapa de calor.
